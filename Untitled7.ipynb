```python
#python program to find how many numbers are havingduplicates
L = [int(i) for i in input().split()]
L2 = []
for i in L:
  if L.count(i)>1:
    L2.append(i)
print(L2)
s=set(L2)
print(s)
print(len(s))
```

```python
n = int(input("enter the elements:"))
L = [int(i) for i in input().split()]
L1 = set(L)
elements = set(range(1,n+1))
print(elements-L1)
```

```python
#set methods
A = {1,2,3,4,5}
B = {4,5,6,7,8}
print("union:",A.union(B))
print("intersection:", A.intersection(B))
print("difference:", A.difference(B))
print("symmetric_difference:", A.symmetric_difference(B))
#add()
A.add(9)
print("After addition A=",A)
#remove
A.remove(3)
print("After removing A=",A)
#pop()
A.pop()
print("After pop A=",A)
#update()
A.update([6,7,8])
print("After updating A=",A)
#discard()
A.discard(5)
print("After discarding A=",A)
#clear
A.clear()
print("After clear A=",A)
```
