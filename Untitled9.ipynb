```python
#write a program that prints frequency of each element in a list in the following
#format [1,2,3,4,1,5,6,1,6,7]     {1:3,2:1,3:1,4:1,5:1,6:2,7:1}
l=[1,2,3,4,1,5,6,1,6,7]
d={}
for i in l:
  if i in d:
    d[i]=d[i]+1
  else:
    d[i]=1
print(d)
```
