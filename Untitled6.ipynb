```python
#TUPLE OPERATIONS
t1 = (10,20,30)
t2 = (40,50)
#concatenation
print("conatenation:",t1+t2)
#repeatation
print("repeatation:",t1*2)
#length
print("length:",len(t1))
#slicilng
print(t1)
#membership
print("20 in tuple:", 20 in t1)
print("60 in tuple:",60 in t1)
#iteration
print("iteration:")
for x in t1:
  print(x)
```

```python
#methods count index
t = (2,5,7,9,5,8)
print(t.count(5))
print(t.index(9))
```

```python
#built in functions in tuple
t = (2,5,7,8,9,3)
print(len(t))
print(max(t))
print(min(t))
print(sum(t))
print(tuple(sorted(t)))
```
