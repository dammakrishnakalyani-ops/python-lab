```python
#1 built in operations
# concatenation
l1 = [1,3,5,"HELLO"]
l2 = [44,55]
print(l1+l2)
#repeatation
print(l1*2)
#indexing
print(l1[1])
#slicing
print(l1[0:3])
#membership
if 44 in l2:
  print("yes")
 # iteration
for i in l1:
  print(i)
```

```python
#2 built in functions
l1 = [3,6,7,2,88,67]
#len()
print(len(l1))
#sum()
print(sum(l1))
#min()
print(min(l1))
#max()
print(max(l1))
#reverse()
print(list(reversed(l1)))
#sorted()
print(sorted(l1))

```

```python
# 3 built in methods of list
l1 = [3,5,7,"kalyani"]
l2 = [2,6,"k"]
#append
l1.append("krishna")
print(l1)
#extend
l2.extend(l1)
print(l2)
#insert
l2.insert(3,"l")
print(l2)
#pop
l2.pop(3)
print(l2)
#remove
l1.remove(5)
print(l1)
# count
print(l1.count(2))
#reverse
l1.reverse()
print(l1)
#sort
l3 = [30,60,7,1,4,678,11,44]
l3.sort()
print(l3)
```

```python
 #program that prints sum of list elements without using sum()
l1 = [3,5,7,35]
sum = 0
for i in l1:
  sum = sum +i
print(sum)
```

```python
#print list elements at odd index
l1 = [4,5,7,8,"kalyani"]
for i in range(1,len(l1),2):
  print(l1[i])
```

```python
#print the even elements
l1 = [2,4,6,8,7,3,9]
for i in l1:
  if i%2==0:
    print(i)
```
